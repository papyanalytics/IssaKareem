```python
#clean the space in the column heading
data.columns = ['Date', 'Medication Revenue', 'Lab Cost', 'Consultation Revenue',
       'Doctor Type', 'Financial Class', 'Patient Type', 'Entry Time',
       'Post-Consultation Time', 'Completion Time', 'Doctor', 'Patient ID']
```

```python
data.columns = data.columns.str.strip()
data.info()
```

```python
data.head()
    
```

```python
#Cleaning Function
def clean_text(x):
    if pd.isnull(x):
        return x  # or return 0 if you prefer to fill NaNs with 0
    x = str(x)  # convert to string first
    x = x.replace('$', '')
    x = x.replace('_', '')
    x = x.replace('-', '')
    x = x.replace(',', '')
    x = x.replace(' ', '')
    x = pd.to_numeric(x, errors='coerce')
    return x
```

```python
clean_text('$1,183.22')
```

```python
data['Medication Revenue'] = data['Medication Revenue'].apply(clean_text)
data.head()
```

```python
data['Lab Cost'] = data['Lab Cost'].apply(clean_text)
data['Consultation Revenue'] = data['Consultation Revenue'].apply(clean_text)
data.info()
```

```python
data = data.fillna(0)
```

```python
data['Total Revenue'] = data['Medication Revenue']+data['Lab Cost']+data['Consultation Revenue']
```

```python
data.head()
```

```python
data.head()
```

```python
# create a quick get hour function
def get_hour(x):
    x=pd.to_datetime(x)
    x = x.hour
    return x
```

```python
data['Entry Hour'] = data['Entry Time'].apply(get_hour)
```

```python
data.head()
```

```python
# create entry hour
data['Entry Hour'] = data['Entry Time'].apply(get_hour)
```

```python
# create other hours
data['Post Consultation Hour'] = data['Post-Consultation Time'].apply(get_hour)
data['Completion Hour'] = data['Completion Time'].apply(get_hour)
data.head()
```

## Visualize Our Results



The organization wants to see a heatmap of when the clinic receives the most patients and revenue

```python
#create days of the week
data['Date'] = pd.to_datetime(data['Date'])
data['dayofweek'] = data['Date'].dt.dayofweek
data['weekday name'] = data['Date'].dt.strftime('%A')
data.head()
```

```python
# import visual libraries
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('dark_background')
```

```python
data.groupby('weekday name')['Total Revenue'].sum().sort_values()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=data, x='weekday name', y='Total Revenue',showfliers=False)
```

```python
plt.figure(figsize=(10,5))
sns.violinplot(data=data, x='weekday name', y='Total Revenue')
```

```python
# create pivot table
plt.figure(figsize=(10,5))
pivot = pd.pivot_table(data, index='Entry Hour', columns='weekday name',values='Total Revenue',aggfunc='sum')
pivot.columns = ['Mon','Tue','Weds','Thurs','Fri','Sat','Sun']
sns.heatmap(pivot,cmap='coolwarm')
```

```python
data.head()
```

```python
# create pivot table
plt.figure(figsize=(10,5))
pivot = pd.pivot_table(data, index='Entry Hour', columns='weekday name',values='Patient ID',aggfunc='count')
pivot.columns = ['Mon','Tue','Weds','Thurs','Fri','Sat','Sun']
sns.heatmap(pivot,cmap='seismic')
```

```python
# create pivot table
plt.figure(figsize=(10,5))
pivot = pd.pivot_table(data, index='Entry Hour', columns='weekday name',values='Total Revenue',aggfunc='sum')
pivot.columns = ['Mon','Tue','Weds','Thurs','Fri','Sat','Sun']
sns.heatmap(pivot,cmap='twilight')
```
